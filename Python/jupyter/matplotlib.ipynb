```python
from matplotlib import pyplot as plt
from matplotlib import style

style.use('ggplot')
x = [1,4,7]
y = [34,11,21]
x2 = [3,7,9]
y2 = [9,13,17]

plt.plot(x,y)
plt.plot(x2,y2)

plt.show()
```

```python
from matplotlib import pyplot as plt

x = [1,4,7]
y = [34,11,21]
x2 = [3,7,9]
y2 = [9,13,17]

plt.plot(x,y, label = "first")
plt.plot(x2,y2, label = "second")
plt.legend()

plt.show()

```

```python
from matplotlib import pyplot as plt

x = [1,4,7]
y = [34,11,21]
x2 = [3,7,9]
y2 = [9,13,17]

plt.plot(x,y, label = "first", linewidth= 6)
plt.plot(x2,y2, label = "second")
plt.legend()

plt.show()
```
