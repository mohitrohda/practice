```python

import spacy
```

```python
nlp = spacy.load("en_core_web_sm")

doc = nlp("Dr. Strange loves pab bhaji of mumbai. Hulk loves chat of delhi")

for sentences in doc.sents:
    print(sentences)
    
```

```python
for sentences in doc.sents:
    for word in sentences:
        print(word)
```
